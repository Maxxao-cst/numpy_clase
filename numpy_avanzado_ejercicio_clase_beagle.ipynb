## Ciencia de datos en imagen 🐶

```python
import numpy as np
#creación de graficas, matplolib más su modulo ".pyplot"
import matplotlib.pyplot as plt

#lector de la imagen para dimesiones (1024, 1024, 3), alto y ancho, canal de colores activos 3 = RGB
img = plt.imread('beagle.jpg')

img.shape
```

```python
#img[10,10,0], el valor 2 hace referencia al canal del RGB, el valor 1 a la cordenada "Y" y el valor 0 a cordenada "X"
rojo = img[10,10,0]
verde = img[10,10,1]
azul = img[10,10,2]

#el resultado es la saturacion/intensidad del color del pixel en esa cordenada
#aqui la cordenada es la misma solo cambia la presencia de color...
# es decir en las cordenadas [10,10] = rojo = 22 / verde = 23 / azul = 5
print ("rojo", rojo)
print ("verde", verde)
print ("azul", azul)
```

```python
print ('Dimensiones: ', img.shape)
print ('Pixel: ', img [10,10])

#mostrar una imagen en algun momento
plt.imshow(img)
#mostrar ejes de imagen, 'off', no mostrarlos, 'on', mostrarlos
plt.axis('on')
#mostrar la imagen
plt.show()
```

```python
#canal minimo de RGB
print ('Valor Minimo:', img.min())
#canal maximo de RGB
print ('Valor Maximo:', img.max())
#color predominante
print ('Media: ', img.mean())
```

```python
#[:,:,0] = cordenadas Null, canal 0 'recuerda que en listados son del 0 al ...'
plt.imshow(img[:,:,0], cmap = 'Reds') #cmap = 'Reds' = color_map = a rojos
plt.colorbar()
plt.title('Canal Rojo')
plt.show()

#canal minimo de RGB
print ('Valor Minimo:', img.min())
#canal maximo de RGB
print ('Valor Maximo:', img.max())
#color predominante
print ('Media: ', img.mean())
```

```python
plt.imshow(img[:,:,1], cmap = 'Greens')
plt.colorbar()
plt.title('Canal Verde')
plt.show()
```

```python
plt.imshow(img[:,:,2], cmap = 'Blues')
plt.colorbar()
plt.title('Canal Azul')
plt.show()
```

```python
canales = ['Rojos', 'Verdes', 'Azules']

media_rojo = img[:,:,0].mean()
media_verdes = img[:,:,1].mean()
media_azules = img[:,:,2].mean()

valores = [media_rojo, media_verdes, media_azules]

plt.bar(canales, valores)
plt.ylabel('Intensidad Promedio')
plt.title ('Color promedio de la imagen')
```
